Task 2 view the distrubution of the various features in the data set and calculate thier central tendecies

we will now look at the distribution of the various features in the data set

we will aslo calculate appropriate measures of central tendency for the features

```python
# Create the histogram of the "Fixed acidity" feature
plt.figure(figsize=(10,6))
sns.histplot(data=data, x = "fixed acidity" , color = 'orange',
             edgecolor = 'c', alpha = 0.5, bins =5)
plt.title("Histogram of the Fixed Acidity", fontweight ='bold')
plt.xlabel("Fixed Acidity")
plt.ylabel("Count")
plt.show()
```

Observations

it histogram is normally distributed

the maximum count of values for fixed acidity line between  6 to 8

lets see the measures of central tendency in working

1. Mean
2. Median

3.Mode

```python
data['fixed acidity'].mean()
```

```python
data['fixed acidity'].median()
```

```python
data['fixed acidity'].mode()
```

```python
# Create the histrogram of the "Fixed acidity " features and also the mean and median
plt.figure(figsize=(9,6)   )
sns.histplot(data=data, x = "fixed acidity" , color = 'orange',
             edgecolor = 'c', alpha = 0.5, bins =5)
plt.title("Histogram of the Fixed Acidity", fontweight ='bold')
plt.xlabel("Fixed Acidity")
plt.ylabel("Count")
plt.vlines(data['fixed acidity']. mean(),ymin = 0, ymax= 4000, color = 'red', label = 'Mean')
plt.vlines(data['fixed acidity']. median(),ymin = 0, ymax= 4000, color = 'Blue',label = 'Median')
plt.legend()
plt.show()
```

Observation

we can see that mean and medina are clear representative of the data,
mean and median are very close to the each other, we can choose the parameters say mean as measure of central tendency

```python
# lets get the output of the
plt.figure(figsize=(10,6))
sns.histplot(data['volatile acidity'] , color = 'g',
             edgecolor = 'g', alpha = 0.5, bins =5)
plt.title("Histogram of the volatile Acidity", fontweight ='bold')
plt.xlabel("Fixed Acidity")
plt.ylabel("Density")
plt.vlines(data['volatile acidity']. mean(),ymin = 0, ymax= 4000, color = 'red', label = 'Mean')
plt.vlines(data['volatile acidity']. median(),ymin = 0, ymax= 4000, color = 'Blue',label = 'Median')
plt.legend()
plt.show()
```

```python
# now making the dist plot for the volatile Acidity


plt.figure(figsize= (8,6))
sns.distplot(data["volatile acidity"],color='blue')
plt.title("Distribution of volatile acidity", fontweight ='bold')
plt.xlabel("volatile acidity")
plt.ylabel('Density')
plt.show()
```

Observation

The Above plot shows the normal distribution


the normal distribution is described by the mean and standard deviation.

the normal distribution is often referred to as  a "bell curve " because of its shape.

the median and mean are equal
it has one mode
it is symmetric , meaning it decreased the same amount on the left and the right of the centre.
